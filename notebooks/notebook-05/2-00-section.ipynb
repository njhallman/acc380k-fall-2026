### A slice includes the start and stops before the end

`text[start:stop]` includes the start position and excludes the stop position. This is sometimes called a **half-open** interval. An omitted endpoint means “continue to that edge.” A slice beyond an edge stops safely; asking for one nonexistent position would raise an error.

A long filing is still text that we can search and slice. We will use [Tesla's 2025 Form 10-K, filed January 29, 2026](https://www.sec.gov/Archives/edgar/data/1318605/000162828026003952/tsla-20251231.htm).

````python
#@title Load Tesla's 2025 10-K as HTML (supplied setup) { display-mode: "form" }
sec_contact = "nicholashallman@utexas.edu" #@param {type:"string"}

import requests
import importlib
import subprocess
import sys
from importlib.metadata import PackageNotFoundError, version
from IPython.display import HTML, display

tesla_url = "https://www.sec.gov/Archives/edgar/data/1318605/000162828026003952/tsla-20251231.htm"
_tesla_edgartools_version = "5.58.0"
# Keep cleaned HTML separate from earlier plain-text and Markdown caches.
_tesla_html_cache = globals().get("_tesla_html_cache", {})
_tesla_cache_key = tesla_url + "|html-v1|" + _tesla_edgartools_version

def _ensure_tesla_parser():
    try:
        installed_version = version("edgartools")
    except PackageNotFoundError:
        installed_version = ""
    if installed_version != _tesla_edgartools_version:
        if "edgar" in sys.modules:
            raise RuntimeError("An earlier EdgarTools version is already loaded; restart the session before setup.")
        print("Installing the tested EdgarTools version; this may take a minute.")
        subprocess.run(
            [sys.executable, "-m", "pip", "install", "--quiet", "--disable-pip-version-check",
             "edgartools==" + _tesla_edgartools_version],
            check=True, timeout=180, capture_output=True, text=True,
        )
        importlib.invalidate_caches()
    from edgar.documents import parse_html
    return parse_html

def _filing_html(document):
    from bs4 import BeautifulSoup, Comment, Doctype
    soup = BeautifulSoup(document.root.html(), "html.parser")
    # Keep display-only content, with no scripts, styles, or external loads.
    for tag in soup.find_all(["head", "script", "style", "iframe", "object", "embed", "img", "link", "meta", "form", "input", "button", "svg", "math"]):
        tag.decompose()
    for item in soup.find_all(string=lambda value: isinstance(value, (Comment, Doctype))):
        item.extract()
    allowed = {"p", "div", "span", "h1", "h2", "h3", "h4", "h5", "h6", "table", "thead", "tbody", "tfoot", "tr", "th", "td", "caption", "b", "strong", "em", "i", "u", "s", "sup", "sub", "br", "hr", "ul", "ol", "li"}
    for tag in soup.find_all(True):
        if tag.name not in allowed:
            tag.unwrap()
        else:
            tag.attrs = {key: value for key, value in tag.attrs.items()
                         if tag.name in {"th", "td", "tr"} and key in {"colspan", "rowspan", "align"}}
    return str(soup).strip()

def _filing_preview(text):
    if not text:
        return ""
    from bs4 import BeautifulSoup
    soup = BeautifulSoup(text, "html.parser")
    # One complete, short cover-page table; never a fabricated passage.
    for table in soup.find_all("table"):
        if "Trading Symbol" in table.get_text():
            return str(table)
    return ""

def _load_tesla_text(contact):
    if _tesla_cache_key in _tesla_html_cache:
        print("Using the full filing HTML already downloaded in this session.")
        return _tesla_html_cache[_tesla_cache_key]
    if not contact.strip() or "@" not in contact:
        print("No filing text loaded. Enter an SEC email contact and rerun setup.")
        return ""
    try:
        response = requests.get(
            tesla_url,
            headers={"User-Agent": "ACC380K classroom text example " + contact.strip()},
            timeout=(5, 15),
        )
        response.raise_for_status()
        if "html" not in response.headers.get("Content-Type", "").lower():
            raise ValueError("The SEC response was not HTML.")
        # This filing is UTF-8. Retain HTML table spans and paragraph tags.
        parse_html = _ensure_tesla_parser()
        text = _filing_html(parse_html(response.content.decode("utf-8")))
        if len(text) < 100000 or "Tesla" not in text or "2025" not in text:
            raise ValueError("The response did not look like the full Tesla filing.")
        _tesla_html_cache[_tesla_cache_key] = text
        print("Downloaded the full 2025 Tesla 10-K and prepared its HTML.")
        return text
    except Exception as error:
        print("Filing setup unavailable:", type(error).__name__)
        if isinstance(error, RuntimeError):
            print(str(error))
        print("No filing text loaded. Check the contact and connection, then rerun setup. Restart the session if a different EdgarTools version was already loaded.")
        return ""

tesla_text = _load_tesla_text(sec_contact)
filing_preview = _filing_preview(tesla_text)
print("Characters loaded:", len(tesla_text))
````

## Exercise — Write a context-extraction function

> **Your turn:** Work on your own or with your table before we work through this together.

Complete `extract_context(text, substring, context_length)`. The inputs are the text to search, the substring to find, and the number of characters to include **on each side** of the match. Spaces count as characters.

Your function must:

- begin with a nonempty docstring;
- find the **first exact, case-sensitive match**; unlike the Tesla example, do not lowercase the inputs;
- return `"Substring not found."` if there is no match;
- otherwise return the entire matched substring with up to `context_length` characters before it and after it, preserving the original text;
- stop at the beginning or end if there are fewer characters available; and
- return the result without printing it.

Assume `substring` is nonempty and `context_length` is a nonnegative integer. You do not need to validate those inputs. A context length of 0 returns just the matched substring.

| Call | Returned string |
|---|---|
| `extract_context("xxCaTyy", "CaT", 1)` | `"xCaTy"` |
| `extract_context("cat---cat", "cat", 2)` | `"cat--"` |
| `extract_context("xxCaTyy", "cat", 1)` | `"Substring not found."` |
| `extract_context("red blue", "blue", 0)` | `"blue"` |

Replace the starter `return None` with your code, then run the collapsed check.

<details><summary>Hint after your first attempt</summary>

Handle a position of `-1` before calculating the slice. A position of 0 is a real match. If your calculated start is negative, use an `if` statement to change it to 0; otherwise Python would count backward from the end. An oversized stop already stops safely at the end of a string.

</details>


````python
def extract_context(text, substring, context_length):
    """ The best docstring ever"""
    pos = text.find(substring)

    if pos == -1:
        return "Substring not found."
    else:
        start = max(pos - context_length,0)
        end = pos + context_length + len(substring)
        return text[start:end]

    return None
````

## Collection-reading card

- string/list/tuple position: `items[0]`; last item: `items[-1]`
- slice: `items[start:stop]` with an excluded stop
- size and membership: `len(items)`, `value in items`
- first text match: `text.find(phrase)`; `-1` means not found
- list change: `items.append(value)` or `items[position] = value`
- flat list copy: `new_items = items.copy()`
- dictionary read/update: `record[key]`, `record[key] = value`; key test: `key in record`
- unpack a pair in order: `first, second = pair`
- return a tuple: `return first, second`
