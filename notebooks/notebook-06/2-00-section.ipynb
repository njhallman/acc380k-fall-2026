## Visit one item at a time

A `for` loop runs the same indented code once for each item in a collection. Read `for city in cities:` in the next cell as "for each city in cities."

- `cities` is the list we visit.
- `city` is the **loop variable**. Python assigns it one item, runs the indented **body**, then moves on to the next item.
- Each trip through the body is one **iteration**.

The colon ends the header, and indentation marks the body. The unindented `print("Finished.")` is outside the loop, so it runs once, after the last iteration. (This `in` means "visit each item of." It is different from the membership question `"Austin" in cities` from Notebook 5.)

**Predict:** What will the cell print, in order? How many times does each `print` run?


````python
cities = ["Austin", "Dallas", "Houston"]

for city in cities:
    print(city)

print("Finished.")
````

## Keep a running total

Printing each number is easy. Adding them up takes memory: after each iteration, we need to know the total so far.

That memory is an **accumulator**: a variable created **before** the loop and updated **inside** it. In the next example:

- `value` is the loop variable. The loop assigns it 4, then 1, then 3.
- `running_total` is the accumulator. It starts at 0, and each iteration adds the current `value` to it.

In `running_total = running_total + value`, Python computes the right side with the old total, then stores the result back in `running_total`.


### Predict, then watch the total

Starting at 0 and adding `[4, 1, 3]` one number at a time, what is `running_total` after the **second** iteration? Commit to a number.


### Trace `value` and `running_total` before you run

Each row is one iteration. `value` is the current item (the loop variable). The other two columns follow the accumulator just before and just after `running_total = running_total + value`. The first row is done; fill in the rest on paper, then run the next cell and compare.

| Iteration | `value` (current item) | `running_total` before addition | `running_total` after addition |
|---:|---:|---:|---:|
| 1 | 4 | 0 | 4 |
| 2 | 1 |  |  |
| 3 | 3 |  |  |


````python
values = [4, 1, 3]
running_total = 0

for value in values:
    print("Current value:", value)
    print("Total before:", running_total)
    running_total = running_total + value
    print("Total after:", running_total)

print("Final total:", running_total)
````

## `range`: count positions

`range` produces whole numbers for a loop to visit. Like a slice, it **includes the start and excludes the stop**:

- `range(stop)` counts 0, 1, …, up to `stop - 1`.
- `range(start, stop)` counts from `start` up to, but not including, `stop`.
- If the stop is at or below the start, the range is empty.

The next cell wraps each range in `list(...)` so we can see all its numbers at once. A loop can use `range(...)` directly.

**Predict:** Write down each list before running. A four-item list has positions 0 through 3. Which of these ranges supplies exactly those positions?


````python
print("range(4):", list(range(4)))
print("range(1, 4):", list(range(1, 4)))
print("range(0):", list(range(0)))
````

### Use positions to read list items

Usually `for amount in amounts:` is all you need. When you also need each item's position, `range(len(amounts))` supplies every valid one: 0 through `len(amounts) - 1`. Remember that `len(amounts)` is the **number of items**, not the last position.

Below, `index` is the position and `amounts[index]` is the value stored there. Predict both numbers on each line.

````python
def invoice_total(prices, quantities):
    """Return the total of each price times the quantity at the same position."""
    total = 0
    for index in range(len(prices)):
        total = total + prices[index]
    return total
````

## Collect selected items in a list

A list can be an accumulator too. Start with `[]` before the loop and `.append()` each item you want to keep. `.append()` changes the list and returns `None` (Notebook 5), so it goes on its own line.

Here we keep amounts of at least 100. Predict `large_amounts` after each iteration. It grows only when the `if` condition is true.


````python
amounts = [40, 120, 75]
large_amounts = []

for amount in amounts:
    if amount >= 100:
        large_amounts.append(amount)
    print("Current amount:", amount, "Kept so far:", large_amounts)
````

- The loop variable receives one item per iteration. An accumulator, created before the loop, keeps the result of earlier iterations.
- `range(stop)` starts at 0; `range(start, stop)` starts at `start`. Both exclude the stop.

- `mapping.items()` supplies key/value pairs from a dictionary.
