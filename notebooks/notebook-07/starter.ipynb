## Check, run, and check again

Like `if`, `while` uses a Boolean condition, a colon, and an indented body. Unlike `if`, after the body finishes Python comes back and checks the condition again:

1. Check the condition **before** running the body.
2. If it is `True`, run the indented lines, then check again.
3. If it is `False`, skip the body and continue with the next unindented line.

## Things we should have covered earlier: `//` and `%`

> **Catch-up section.** Notebook 2 introduced `+`, `-`, `*`, `/`, and `**`, but two more arithmetic operators belong on that list. We use both in the rest of this notebook, so we add them here.

Think back to long division: 17 divided by 5 is 3 with a remainder of 2. Python has an operator for each piece:

| Operator | Name | Example | Result |
|---|---|---|---:|
| `/` | division (always a `float`) | `17 / 5` | `3.4` |
| `//` | **floor division**: how many whole times it fits | `17 // 5` | `3` |
| `%` | **remainder** (also called *modulo*): what is left over | `17 % 5` | `2` |

````python
print(17 / 5)
print(17 // 5)
print(17 % 5)
````

Three things to remember for now:

- With whole numbers, `//` and `%` give whole numbers (`int`), not `float` values.
- Use them with nonnegative numbers for now. (With negative numbers `//` rounds *down* to the smaller whole number, so `-7 // 2` is `-4`, not `-3`.)
- Dividing by zero is an error for all three operators.

## Steering a loop: `break`, `continue`, and `pass`

So far a loop runs its body for every item (or until its condition is false). Three short statements let you change that:

| Statement | What it does |
|---|---|
| `break` | Leave the loop **now**. Skip all remaining iterations. |
| `continue` | Skip the **rest of this iteration** and go straight to the next one. |
| `pass` | Do **nothing**. It only holds a place where Python requires a statement. |

`break` and `continue` work in both `for` and `while` loops, and they almost always sit inside an `if`.

### `break`: stop when something happens

A client's account holds \$100. We post these withdrawals in order, but must stop at the first one the balance cannot cover. **Predict:** which withdrawals are posted, what prints inside the loop, and what is the final balance?


````python
balance = 100
withdrawals = [30, 50, 40, 20]

for withdrawal in withdrawals:
    if withdrawal > balance:
        print("Stop: cannot cover", withdrawal)
        break
    balance = balance - withdrawal
    print("Posted", withdrawal, "- balance is now", balance)

print("Final balance:", balance)
````

The 20 at the end of the list could have been covered by the remaining balance of 20, but it was never looked at: `break` ended the whole loop at 40. Notice that `print("Final balance:", balance)` is unindented, so it runs once after the loop, whether the loop ended by `break` or by running out of items.

### `continue`: skip this item and keep going

`continue` ends only the current iteration. Here, a zero or negative entry is not a purchase, so we skip it and keep adding the rest. **Predict:** what is `total` at the end?


````python
entries = [40, -15, 25, 0, 10]
total = 0

for entry in entries:
    if entry <= 0:
        print("Skipping", entry)
        continue
    total = total + entry
    print("Added", entry, "- total is now", total)

print("Total:", total)
````

### `continue` in a `while` loop: update first

`continue` jumps straight back to the condition check. In a `while` loop, any line **below** the `continue` is skipped. If that skipped line is the one that makes the condition eventually false, you get an infinite loop. The safe habit: make the change that moves the loop forward **before** the `continue`.

The next cell processes invoice numbers 1 through 6 but skips every third one. Notice how `%` from the earlier section picks them out. **Predict** the printed invoice numbers.


````python
number = 0

while number < 6:
    number = number + 1          # move forward first
    if number % 3 == 0:
        continue                 # skip invoices 3 and 6
    print("Process invoice", number)

print("Done.")
````

### `pass` in a loop body

`pass` is not only for `if`. A loop body cannot be empty either, so `pass` can stand in for a loop body you have not written yet. And because `pass` does nothing, putting it on the last line of a loop body changes nothing.

````python
for tx_id in ["T1", "T2", "T3"]:
    pass                         # the loop body will be written later

print("After the first loop, tx_id is", tx_id)

count = 0
for tx_id in ["T1", "T2", "T3"]:
    count = count + 1
    pass                         # last line of the body: still does nothing

print("Count:", count)
````

The first loop ran three times. Each time, `tx_id` took the next ID and `pass` did nothing, so after the loop `tx_id` still holds the last ID, `"T3"`. In the second loop, deleting the `pass` line would not change the output.

**Careful with `while`:** a `while` loop whose body is only `pass` never changes anything, so if its condition starts out `True` it never becomes `False`. That is an infinite loop, so do not leave `pass` as the whole body of a `while` loop.

## List comprehensions: build a list in one line

In Notebook 6 you built lists with a loop: start with `[]`, then `.append()` inside a `for`. That pattern is so common that Python has a one-line version called a **list comprehension**.

The two cells below build the same list: each amount of cents converted to whole dollars with `//`. **Predict:** what do both print?


````python
amounts_cents = [1275, 805, 99, 2500]

# The loop you know:
dollars_loop = []
for cents in amounts_cents:
    dollars_loop.append(cents // 100)

# The same list as a comprehension:
dollars_comp = [cents // 100 for cents in amounts_cents]

print(dollars_loop)
print(dollars_comp)
print(dollars_loop == dollars_comp)
````

### The two parts of a comprehension

```text
[ cents // 100   for cents in amounts_cents ]
  ^^^^^^^^^^^^   ^^^^^^^^^^^^^^^^^^^^^^^^^^
  what to keep   the loop header, without the colon
```

Read it as "the list of `cents // 100`, for each `cents` in `amounts_cents`." The `for` part is the loop header you already know, and the expression in front replaces `.append(...)`. The brackets `[...]` tell Python to collect the results into a new list, in the original order. The original list is not changed.

### Add a condition

An `if` at the end keeps only the items that pass, just like the `if` before `.append()` in Notebook 6's "collect selected items" pattern. **Predict** each list before running.


````python
amounts_cents = [1275, 805, 99, 2500]

# Dollars for amounts of at least $10.00 (1000 cents):
big_dollars = [cents // 100 for cents in amounts_cents if cents >= 1000]
print(big_dollars)

# Keep the items themselves (nothing is changed), only if they pass a test:
even_numbers = [number for number in range(1, 11) if number % 2 == 0]
print(even_numbers)

# A condition that keeps nothing gives an empty list:
print([cents for cents in amounts_cents if cents > 99999])
````

### The same list as a loop

The comprehension with a condition is a compact version of this loop. Match each piece:

```python
big_dollars = []                       # the new list: the [ ] brackets
for cents in amounts_cents:            # for cents in amounts_cents
    if cents >= 1000:                  # if cents >= 1000
        big_dollars.append(cents // 100)   # cents // 100 (the item kept)
```

**Order to remember in the comprehension:** *what to keep* → `for` header → optional `if`.

### When to use which

- Use a **comprehension** when the whole job is "make a new list from this collection," optionally keeping only some items.
- Use a regular **`for` loop** when you also need a running total or a count, a `print`, a `break` or `continue`, or more than one step per item. A comprehension cannot use `break` or `continue`.
- If a comprehension makes you pause to read it, write the loop. Clear beats short.

## Putting it together: monetary unit sampling

**Monetary unit sampling (MUS)**, also called dollar-unit sampling, is one way auditors choose which transactions to test. A quick refresher:

- Every **dollar** in the population is a sampling unit. A \$300 transaction holds 300 dollar units; a \$15 transaction holds 15.
- The auditor picks a **sampling interval**, such as every 150th dollar. It is often set by dividing the population total by the sample size wanted.
- A **random start** from 1 to the interval picks the first dollar. After that, every interval-th dollar is selected: the random start, then the random start + 150, then + 300, and so on, until the selections pass the population total.
- Each selected dollar points to the **transaction** that holds it, and that transaction is tested. Larger transactions hold more dollars, so they are more likely to be picked. A transaction larger than the interval is always picked, sometimes more than once.

### A random start with `random`

Python comes with a `random` module for making random numbers; there is nothing to install. `import random` loads it, and `random.randint(1, interval)` returns a random whole number from 1 to `interval`, **including both ends**.

## Recap

- **`while`** repeats while a condition is `True`. Something in the body must change so the condition can become `False`.
- **`//`** gives how many whole times one number fits in another; **`%`** gives what is left over. Together they split dollars and cents, find even and odd, pick every Nth item, and size batches.
- **`break`** leaves the loop; **`continue`** skips to the next iteration (in a `while` loop, update first); **`pass`** does nothing and only holds a place.
- A **list comprehension** `[expression for item in items if condition]` builds a new list in one line. Use a regular loop when you need more than a new list.
- **Monetary unit sampling** fits a `while` loop inside a `for` loop: the `for` loop walks the transactions with a running total, and the `while` loop takes every selected dollar inside the current one.
