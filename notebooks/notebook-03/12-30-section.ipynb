## Parameters, arguments, local names, and `return`

Now define a function that receives an input and sends a value back:

```python
def make_cheer(team):
    message = "Go " + team + "!"
    return message
```

- `team` is a **parameter**: a name in the function definition.
- `"Longhorns"` in `make_cheer("Longhorns")` is an **argument**: the value supplied by the call. For that call, Python uses `"Longhorns"` as the value of `team`.
- `message` is a **local name**: it is created inside the function and is available only while that call is running.
- `return` sends the value of `message` back to the calling line.


````python
def make_cheer(team):
    message = "Go " + team + "!"
    return message
````

## Defaults, keyword arguments, and docstrings

A parameter can have a default value. In the next definition, `greeting="Welcome"` means that a call may omit `greeting`; Python will use `"Welcome"`.

Arguments written without parameter names are **positional arguments**. Python matches them to parameters from left to right. A **keyword argument** supplies a value by naming its parameter in the call.

A call may use both forms. Python matches the positional arguments first, from left to right; any keyword arguments then supply values to the parameters they name. In `greet("Avery", greeting="Good afternoon")`, the positional argument fills `name`, and the keyword argument fills `greeting`. Positional arguments must come before keyword arguments, and the same parameter cannot receive two values. When every argument is named, the keyword arguments may appear in either order.

A triple-quoted string used as the first statement inside a function is its **docstring**. A docstring describes what the function does, and tools such as `help()` can display it.

Before running the next cell, predict all five displayed lines. For each call, identify whether it uses the default value, positional arguments, keyword arguments, or a mixture of positional and keyword arguments.


````python
def greet(greeting="Welcome", name):
    """
    Return a greeting for one name.

    Here is another line of docs
    """
    return greeting + ", " + name + "!"

print(greet("Avery"))
print(greet("Avery", "Hello"))
print(greet("Avery", greeting="Good afternoon"))
print(greet(name="Avery", greeting="Good morning"))
print(greet(greeting="Good morning", name="Avery"))
````

````python
greet("Good morning", "Avery")
````

````python
help(greet)
````

### Compare the calls

The first call omits `greeting`, so Python uses the default. The second call supplies both values by position. The third call supplies `name` by position and `greeting` by keyword. The final two calls name both parameters, so reversing their order does not change which value each parameter receives.

## Exercise — Use a default, a docstring, and a keyword call

> **Your turn:** Work on your own or with your table before we work through this together.

Complete both functions in the next cell.

`calculate_trip_cost(distance_miles, miles_per_gallon, fuel_price_per_gallon=3.50)` must:

- change the supplied default from `None` to `3.50`;
- begin with a nonempty docstring;
- calculate `gallons_needed` as a local value; and
- return the trip cost without printing it.

`cost_for_standard_trip()` must return the result of calling `calculate_trip_cost()` with the keyword arguments `distance_miles=240` and `miles_per_gallon=30`. Omit `fuel_price_per_gallon` so the call uses its default.

The collapsed check tries the default, an explicit fuel price, a mixed positional-and-keyword call, and the standard-trip helper. You do not need to open or understand the check's internal code.


````python
def calculate_trip_cost(distance_miles, miles_per_gallon, fuel_price_per_gallon=3.50):
    """
    Calculates the cost of a trip based on miles, gallons of fuel per mile and the cost of fuel.
    """
    gallons_needed = distance_miles / miles_per_gallon
    return gallons_needed * fuel_price_per_gallon


def cost_for_standard_trip():
    return calculate_trip_cost(240, 30)
````

````python
#@title Check your work { display-mode: "form" }

import contextlib
import inspect
import io
import math
import sys

trip_function = globals().get("calculate_trip_cost")
standard_function = globals().get("cost_for_standard_trip")

def close_enough(actual, expected):
    try:
        return math.isclose(
            actual,
            expected,
            rel_tol=1e-9,
            abs_tol=1e-9,
        )
    except (TypeError, ValueError):
        return False

checks = {
    "Both functions are defined": (
        callable(trip_function)
        and callable(standard_function)
    ),
    "The signature uses the required names and default": False,
    "cost_for_standard_trip takes no arguments": False,
    "The trip-cost function has a nonempty docstring": False,
    "gallons_needed is calculated locally": False,
    "Default, explicit, and changed inputs are correct": False,
    "Both functions return without printing": False,
    "The helper uses keywords and the default price": False,
}
first_error = None

if callable(trip_function):
    try:
        parameters = list(
            inspect.signature(trip_function).parameters.values()
        )
        checks[
            "The signature uses the required names and default"
        ] = (
            [parameter.name for parameter in parameters]
            == [
                "distance_miles",
                "miles_per_gallon",
                "fuel_price_per_gallon",
            ]
            and parameters[0].default
            is inspect.Parameter.empty
            and parameters[1].default
            is inspect.Parameter.empty
            and parameters[2].default == 3.50
        )

        docstring = inspect.getdoc(trip_function)
        checks[
            "The trip-cost function has a nonempty docstring"
        ] = isinstance(docstring, str) and bool(docstring.strip())

        captured_gallons = []
        missing_local = object()
        trip_code = getattr(trip_function, "__code__", None)

        def capture_trip_return(frame, event, argument):
            if frame.f_code is trip_code and event == "return":
                captured_gallons.append(
                    frame.f_locals.get(
                        "gallons_needed",
                        missing_local,
                    )
                )
            return capture_trip_return

        displayed = io.StringIO()
        prior_trace = sys.gettrace()
        sys.settrace(capture_trip_return)
        try:
            with contextlib.redirect_stdout(displayed):
                default_actual = trip_function(240, 30)
                explicit_actual = trip_function(100, 25, 4.00)
                mixed_actual = trip_function(
                    60,
                    miles_per_gallon=20,
                    fuel_price_per_gallon=5.00,
                )
                standard_actual = (
                    standard_function()
                    if callable(standard_function)
                    else None
                )
        finally:
            sys.settrace(prior_trace)

        checks[
            "gallons_needed is calculated locally"
        ] = (
            len(captured_gallons) >= 3
            and all(
                close_enough(actual, expected)
                for actual, expected in zip(
                    captured_gallons[:3],
                    (8.0, 4.0, 3.0),
                )
            )
        )

        checks[
            "Default, explicit, and changed inputs are correct"
        ] = all(
            (
                close_enough(default_actual, 28.0),
                close_enough(explicit_actual, 16.0),
                close_enough(mixed_actual, 15.0),
                close_enough(standard_actual, 28.0),
            )
        )
        checks[
            "Both functions return without printing"
        ] = displayed.getvalue() == ""
    except Exception as error:
        first_error = type(error).__name__ + ": " + str(error)

if callable(standard_function):
    try:
        checks[
            "cost_for_standard_trip takes no arguments"
        ] = not inspect.signature(
            standard_function
        ).parameters
    except Exception as error:
        if first_error is None:
            first_error = (
                type(error).__name__ + ": " + str(error)
            )

    original_trip_function = globals().get(
        "calculate_trip_cost"
    )
    recorded_calls = []
    sentinel = object()

    def record_trip_call(*args, **kwargs):
        recorded_calls.append((args, kwargs))
        return sentinel

    globals()["calculate_trip_cost"] = record_trip_call
    try:
        with contextlib.redirect_stdout(io.StringIO()):
            helper_result = standard_function()
        checks[
            "The helper uses keywords and the default price"
        ] = (
            recorded_calls
            == [
                (
                    (),
                    {
                        "distance_miles": 240,
                        "miles_per_gallon": 30,
                    },
                )
            ]
            and helper_result is sentinel
        )
    except Exception as error:
        if first_error is None:
            first_error = (
                type(error).__name__ + ": " + str(error)
            )
    finally:
        globals()["calculate_trip_cost"] = (
            original_trip_function
        )

for label, passed in checks.items():
    print(label + ":", "PASS" if passed else "NOT YET")
if first_error is not None:
    print("First error:", first_error)
print("Overall:", "PASS" if all(checks.values()) else "NOT YET")
````

## Function-reading card

```python
def function_name(parameter):
    local_value = expression_using(parameter)
    return local_value

saved_result = function_name(argument)
```

- In a definition, `parameter=default_value` gives the parameter a default.
- In a call, `parameter=argument` supplies a keyword argument by name.
- A docstring is the first string inside the function body and explains what the function does.

Ask three trust questions:

2. Does a changed input still behave correctly?
3. If a call omits an optional argument, does the expected default apply?
