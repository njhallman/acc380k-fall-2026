## `if` and `else` choose a path

Python evaluates the condition after `if`.

- If the condition is `True`, Python runs the indented block under `if`.
- If the condition is `False`, Python skips the `if` block. When an `else` is present, Python runs its indented block.
- After the selected block finishes, execution continues with the next unindented statement.

The colon and indentation are part of the syntax. They show which statements belong to each path.

An `else` is optional. Without one, Python skips the `if` block when the condition is `False` and continues with the next unindented statement. If this happens inside a function and no later `return` runs, reaching the end of the function returns `None`.

## Separate decisions and one multiway decision

Two separate `if` statements ask two independent questions. Both indented blocks can run when both conditions are true.

An `if`/`elif`/`else` chain represents one multiway choice. Python checks its conditions from top to bottom. It runs the **first true branch** and skips the rest of the chain.

This chain often does the same job as nested `IF` functions in Excel: it selects one result from several alternatives. Python also allows an `if` inside another branch. Use nesting when a second question should be asked only within an earlier path. For one ordered set of alternative thresholds, an `elif` chain is usually easier to read.

Now the same thresholds select one label. Predict the label for 120, then run the cell. Change the final call to `size_label(75)` and rerun it.


````python
def size_label(amount):
    if amount >= 100:
        return "large"
    elif amount >= 50:
        return "medium"
    else:
        return "small"

print(size_label(120))
````

### Compare the structures

For 120, both separate `if` conditions are true, so both independent blocks run. In the chain, `amount >= 100` is the first true condition, so the function returns `"large"` and skips the later branches.

Use separate `if` statements when more than one independent action may be needed. Use an `if`/`elif`/`else` chain when the paths are alternatives and exactly one should be selected.

````python
def expense_review(amount):
    """Return the required review level for one expense amount."""
    if amount < 0:
        return "invalid"
    elif amount >= 1000:
        return "supervisor review"
    elif amount >= 5000:
        return "manager review"
    elif amount > 10000:
        return "director review"
    else:
        return "automatic"
````

````python
expected_labels = {
    7500: "manager review",
    9999: "manager review",
    10000: "director review",
    10001: "director review",
}

for amount, expected in expected_labels.items():
    actual = expense_review(amount)
    print(amount, "— expected:", expected, "— actual:", actual)
````

````python
#@title Check your work { display-mode: "form" }
expense_cases = {
    -1: "invalid",
    0: "automatic",
    1: "automatic",
    999: "automatic",
    1000: "supervisor review",
    1001: "supervisor review",
    4999: "supervisor review",
    5000: "manager review",
    5001: "manager review",
    9999: "manager review",
    10000: "director review",
    10001: "director review",
    12000: "director review",
}

expense_results = {
    amount: expense_review(amount)
    for amount in expense_cases
}
expense_mismatches = {
    amount: (expense_cases[amount], expense_results[amount])
    for amount in expense_cases
    if expense_results[amount] != expense_cases[amount]
}

print("All cases passed:", not expense_mismatches)
if expense_mismatches:
    print("Mismatches shown as input: (expected, actual)")
    print(expense_mismatches)
````

## Exercise — Return a course letter grade

> **Your turn:** Work on your own or with your table before we work through this together.

Write `course_letter_grade(homework_points, case1_points, case2_points, midterm_points)` from scratch.

The four components contain 100 possible course points:

- homework: 25 points;
- Case 1: 20 points;
- Case 2: 30 points; and

Your function must:

- begin with a nonempty docstring;
- calculate the sum as a local value named `total_points`;
- use the published floors below, in an order that makes the first true branch the correct one; and
- return the letter grade without printing it.

| Published floor | Grade |
|---:|:---|
| at least 93 | `"A"` |
| at least 90 | `"A-"` |
| at least 87 | `"B+"` |
| at least 83 | `"B"` |
| at least 80 | `"B-"` |
| at least 77 | `"C+"` |
| at least 73 | `"C"` |
| at least 70 | `"C-"` |
| at least 60 | `"D"` |
| below 60 | `"F"` |

Assume all four inputs are valid course-point amounts within their stated maxima. Do not round `total_points`. These are the published floors before any possible upward curve.

Run the collapsed check when your function is ready. It tests every letter-grade boundary and changes each component separately.


````python
# Write course_letter_grade(homework_points, case1_points, case2_points, midterm_points) here.
````

````python
#@title Check your work { display-mode: "form" }
import contextlib
import inspect
import io

grade_function = globals().get("course_letter_grade")
expected_parameters = [
    "homework_points",
    "case1_points",
    "case2_points",
    "midterm_points",
]
parameters_passed = False
docstring_passed = False
local_total_passed = False
no_print_reference_passed = False
no_round_reference_passed = False
no_displayed_output_passed = False
cases_passed = False

if not callable(grade_function):
    print("Function defined: False")
else:
    print("Function defined: True")
    actual_parameters = list(inspect.signature(grade_function).parameters)
    parameters_passed = actual_parameters == expected_parameters
    docstring_passed = bool(inspect.getdoc(grade_function))
    local_total_passed = (
        "total_points" in grade_function.__code__.co_varnames
    )
    no_print_reference_passed = (
        "print" not in grade_function.__code__.co_names
    )
    no_round_reference_passed = (
        "round" not in grade_function.__code__.co_names
    )

grade_cases = [
    ("93-point A floor", (25, 20, 30, 18), "A"),
    ("just below A", (25, 20, 30, 17.99), "A-"),
    ("90-point A- floor", (25, 20, 30, 15), "A-"),
    ("just below A-", (25, 20, 30, 14.99), "B+"),
    ("87-point B+ floor", (25, 20, 30, 12), "B+"),
    ("just below B+", (25, 20, 30, 11.99), "B"),
    ("83-point B floor", (25, 20, 30, 8), "B"),
    ("just below B", (25, 20, 30, 7.99), "B-"),
    ("80-point B- floor", (25, 20, 30, 5), "B-"),
    ("just below B-", (25, 20, 30, 4.99), "C+"),
    ("77-point C+ floor", (25, 20, 30, 2), "C+"),
    ("just below C+", (25, 20, 30, 1.99), "C"),
    ("73-point C floor", (25, 20, 28, 0), "C"),
    ("just below C", (25, 20, 27.99, 0), "C-"),
    ("70-point C- floor", (25, 20, 25, 0), "C-"),
    ("just below C-", (25, 20, 24.99, 0), "D"),
    ("60-point D floor", (25, 20, 15, 0), "D"),
    ("just below D", (25, 20, 14.99, 0), "F"),
    ("homework changed", (10, 20, 30, 25), "B"),
    ("Case 1 changed", (25, 5, 30, 25), "B"),
    ("Case 2 changed", (25, 20, 15, 25), "B"),
    ("midterm changed", (25, 20, 30, 10), "B"),
]

grade_mismatches = []
captured_output = io.StringIO()
if callable(grade_function) and parameters_passed:
    with contextlib.redirect_stdout(captured_output):
        for case_name, arguments, expected in grade_cases:
            try:
                actual = grade_function(*arguments)
            except Exception as error:
                actual = type(error).__name__ + ": " + str(error)
            if actual != expected:
                grade_mismatches.append(
                    (case_name, sum(arguments), expected, actual)
                )
    no_displayed_output_passed = captured_output.getvalue() == ""
    cases_passed = not grade_mismatches

print("Four parameters in the required order:", parameters_passed)
print("Nonempty docstring:", docstring_passed)
print("Local total_points name:", local_total_passed)
print("No print() call in the function:", no_print_reference_passed)
print("No round() call in the function:", no_round_reference_passed)
print("No output during the checked calls:", no_displayed_output_passed)
print("All grade cases passed:", cases_passed)

if not parameters_passed:
    print("Fix the parameter list before the grade cases can run.")
elif grade_mismatches:
    print("Mismatches shown as case: total, expected, actual")
    for case_name, total, expected, actual in grade_mismatches:
        print(case_name + ":", total, expected, actual)

print(
    "All requirements passed:",
    callable(grade_function)
    and parameters_passed
    and docstring_passed
    and local_total_passed
    and no_print_reference_passed
    and no_round_reference_passed
    and no_displayed_output_passed
    and cases_passed,
)
````

```python
if first_condition:
    return first_result
elif second_condition:
    return second_result
else:
    return fallback_result
```

- Python checks one chain from top to bottom;
- the first true branch wins;
- separate `if` statements are separate decisions;
- place overlapping `>=` thresholds from largest to smallest; and
